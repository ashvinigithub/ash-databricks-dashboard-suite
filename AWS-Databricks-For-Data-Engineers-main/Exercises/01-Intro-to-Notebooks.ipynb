```python
csv_path = '/databricks-datasets/Rdatasets/data-001/csv/ggplot2/diamonds.csv'

diamonds_df = (spark.read
    .option('header', 'true')     # first row is the column header
    .option('inferSchema', 'true') # auto-detect column data types
    .csv(csv_path)
)

result_df = (diamonds_df
    .select('cut', 'color', 'clarity', 'price', 'carat')
    .filter((diamonds_df['cut'] == 'Premium') & (diamonds_df['price'] > 5000))
)

display(result_df)
```
